```python
import pandas as pd
import numpy as np

brfss = "https://raw.githubusercontent.com/zevdavis/trubridge_extern/refs/heads/main/brfss_np.csv"
s1701 = "https://raw.githubusercontent.com/zevdavis/trubridge_extern/refs/heads/main/S1701_pd.csv"

df1 = pd.read_csv(brfss)
df2 = pd.read_csv(s1701)
```

```python
question1 = df1.loc[df1["Question"].str.contains("activities")]
question1
```

```python
question2 = df1.loc[df1["Question"].str.contains("dysthymia")]
question2
```

```python
question3 = df1.loc[df1["Question"].str.contains("emotions")]
question3
```

```python
q1_mean = question1.groupby("Area")["Percent"].mean()
print(q1_mean)
```

```python
q2_mean = question2.groupby("Area")["Percent"].mean()
print(q2_mean)
```

```python
q3_mean = question3.groupby("Area")["Percent"].mean()
print(q3_mean)
```

```python
brfss_means = pd.DataFrame({"activities": q1_mean,
                            "depressive_disorder": q2_mean,
                            "bad_mh": q3_mean})
print(brfss_means)
```

```python
s1701_perc = df2.loc[:, df2.columns.str.contains("!!Percent below poverty level!!Estimate")].iloc[0]
s1701_perc
```

```python
s1701_perc = pd.to_numeric(s1701_perc.str.replace("%", ""))
s1701_perc
```

```python
rename_dict = {col: col.split("!!")[0] for col in s1701_perc.index}
s1701_perc = s1701_perc.rename(rename_dict)
s1701_perc
```

```python
df_perc = brfss_means.join(s1701_perc)
df_perc
```

```python
df_perc = df_perc.rename(columns={0: "und_pov_lev"})
df_perc
```

```python
correlation_matrix = df_perc.corr()
print(correlation_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
from scipy import stats
```

```python
r1, p1 = stats.pearsonr(df_perc["und_pov_lev"], df_perc["activities"])
print("r1:", r1)
print("p-value1:", p1)
```

```python
r2, p2 = stats.pearsonr(df_perc["und_pov_lev"], df_perc["depressive_disorder"])
print("r2:", r2)
print("p-value2:", p2)
```

```python
r3, p3 = stats.pearsonr(df_perc["und_pov_lev"], df_perc["bad_mh"])
print("r3:", r3)
print("p-value3:", p3)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X = df_perc[["und_pov_lev"]]
y1 = df_perc[["activities"]]
y2 = df_perc[["depressive_disorder"]]
y3 = df_perc[["bad_mh"]]
```

```python
model1 = LinearRegression()
model1.fit(X, y1)
model2 = LinearRegression()
model2.fit(X, y2)
model3 = LinearRegression()
model3.fit(X, y3)
```

```python
print("Activities - Coefficient:", model1.coef_)
print("Activities - R^2:", model1.score(X, y1))
print("Depressive Disorder - Coefficient:", model2.coef_)
print("Depressive Disorder - R^2:", model2.score(X, y2))
print("Bad MH - Coefficient:", model3.coef_)
print("Bad MH - R^2:", model3.score(X, y3))
```

```python
plt.scatter(X, y1)
plt.xlabel("Below Poverty Level Rate (%)")
plt.ylabel("Impact of Depression on Activities (%)")
plt.title("Poverty and Activities Model")
plt.plot(X, model1.predict(X), color="red")
plt.show()
```

```python
plt.scatter(X, y2)
plt.xlabel("Below Poverty Level Rate (%)")
plt.ylabel("Reported Depressive Disorders (%)")
plt.title("Poverty and Depressive Disorders Model")
plt.plot(X, model2.predict(X), color="red")
plt.show()
```

```python
plt.scatter(X, y3)
plt.xlabel("Below Poverty Level Rate (%)")
plt.ylabel("Reported Bad Mental Health (%)")
plt.title("Poverty and Bad Mental Health Model")
plt.plot(X, model3.predict(X), color="red")
plt.show()
```
